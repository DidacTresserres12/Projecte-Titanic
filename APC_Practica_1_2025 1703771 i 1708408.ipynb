```python
#2.
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
# Extraer título de 'Name' 
train['Title'] = train['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
test['Title'] = test['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)

# Tamaño de la familia
train['FamilySize'] = train['SibSp'] + train['Parch'] + 1
test['FamilySize'] = test['SibSp'] + test['Parch'] + 1

# Variable binaria: viaja solo o no
train['IsAlone'] = (train['FamilySize'] == 1).astype(int)
test['IsAlone'] = (test['FamilySize'] == 1).astype(int)

#Seleccionamos las variables
numeric_features = ['Age', 'Fare', 'FamilySize']
categorical_features = ['Pclass', 'Sex', 'Embarked', 'Title', 'IsAlone']

#Preproceso con pipeline

# Rellenamos los nulos con la mediana y normalizamos para las numéricas.
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())  # normalización: útil para modelos como SVM o KNN
])

# Para variables categóricas usamos One-Hot Encoding para unificar
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combinamos numericas y categóricas
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

#Una vez hehco el preprocesado lo aplicamos a nuestro entrenamiento
X_train = preprocessor.fit_transform(train)
y_train = train['Survived']

X_test = preprocessor.transform(test)

print("Dimensiones X_train:", X_train.shape)
print(X_train[:5])
```

```python
#3.

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    average_precision_score,
    precision_recall_curve,
    roc_curve,
    roc_auc_score,
    classification_report
)
import matplotlib.pyplot as plt

#entrenemamos nuestro modelo segun nuestra clasificación y selección de variables
# Separar variabless
numeric_features = ['Age', 'Fare', 'FamilySize']
categorical_features = ['Pclass', 'Sex', 'Embarked', 'Title', 'IsAlone']

# Copiar columnas
X = train[numeric_features + categorical_features].copy()
y = train['Survived']

# Convertir numéricas a float y rellenar NaNs
for col in numeric_features:
    X[col] = pd.to_numeric(X[col], errors='coerce')
X[numeric_features] = X[numeric_features].fillna(0)

# Rellenar NaNs en categóricas con la moda
for col in categorical_features:
    X[col] = X[col].fillna(X[col].mode()[0])

# One-hot encoding
X = pd.get_dummies(X, columns=categorical_features, drop_first=True)

# Entrenar el modelo sobre todo el dataset
log_reg = LogisticRegression(max_iter=1000)
log_reg.fit(X, y)

# Predicciones sobre todo el dataset
y_pred = log_reg.predict(X)
y_pred_proba = log_reg.predict_proba(X)[:, 1]

#Calculamos las 3 métricas requeridas
acc = accuracy_score(y, y_pred)
f1 = f1_score(y, y_pred)
avg_prec = average_precision_score(y, y_pred_proba)

print("Accuracy:", acc)
print("F1:", f1)
print("Average Precision:", avg_prec)

#Mostramos la Classification Report
print("\nClassification Report:\n", classification_report(y, y_pred))

# ROC Curve
fpr, tpr, _ = roc_curve(y, y_pred_proba)
plt.plot(fpr, tpr)
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve")
plt.show()

# Precision-Recall Curve
precision, recall, _ = precision_recall_curve(y, y_pred_proba)
plt.plot(recall, precision)
plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.show()
```

```python
#4.

import time
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import f1_score, make_scorer
from sklearn.impute import SimpleImputer

#Miramos que no haya ningun error y preparamos los datos x y y
try:
    X
    y
except NameError:
    from sklearn.datasets import fetch_openml
    data = fetch_openml("titanic", version=1, as_frame=True)
    X = data.data
    y = (data.target == "survived").astype(int)

X = pd.DataFrame(X).reset_index(drop=True)
y = pd.Series(y).reset_index(drop=True)

#Separamos nuestro dataset y asignamos columnas
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

cat_cols = X_train.select_dtypes(include=['object', 'category']).columns
num_cols = X_train.select_dtypes(exclude=['object', 'category']).columns

#Hacemos el preprocesador con pipeline y como habíamos usado antes usamos la media y el más frequente en función del tipo de variable
numeric_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, num_cols),
    ('cat', categorical_transformer, cat_cols)
])

#Definimos los 4 modelos seleccionados
models = {
    'LogisticRegression': Pipeline([
        ('pre', preprocessor),
        ('clf', LogisticRegression(max_iter=2000, random_state=42))
    ]),
    'KNN': Pipeline([
        ('pre', preprocessor),
        ('clf', KNeighborsClassifier())
    ]),
    'SVM_linear': Pipeline([
        ('pre', preprocessor),
        ('clf', SVC(kernel='linear', probability=True, random_state=42))
    ]),
    'SVM_rbf': Pipeline([
        ('pre', preprocessor),
        ('clf', SVC(kernel='rbf', probability=True, random_state=42))
    ])
}

#Aplicamos el CrossValidation y el F1Score
cv = StratifiedKFold(n_splits=3, shuffle=True, random_state=42)
scorer = make_scorer(f1_score)

#Comparamos modelos con los hiperparámetros por defecto
results_default = []
for name, pipeline in models.items():
    try:
        start = time.time()
        scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring=scorer, n_jobs=-1)
        elapsed = time.time() - start
        results_default.append({
            'Model': name,
            'Mean CV F1 (train)': np.mean(scores),
            'Std CV F1': np.std(scores),
            'Train Time (s) (CV)': round(elapsed, 2)
        })
    except Exception as e:
        print(f" Error en model {name}: {e}")
        results_default.append({
            'Model': name,
            'Mean CV F1 (train)': None,
            'Std CV F1': None,
            'Train Time (s) (CV)': None
        })
#Mostramos una tabla que nos permite comparar los parametros con los hiperparámetros por defecto
results_default_df = pd.DataFrame(results_default).sort_values('Mean CV F1 (train)', ascending=False).reset_index(drop=True)
display(results_default_df)

#Seleccionamos los hiperparametros a tener en cuenta de cada modelo
param_grids = {
    'LogisticRegression': {'clf__C': [0.01, 0.1, 0.3, 1, 3, 10], 'clf__solver': ['liblinear', 'lbfgs']},
    'KNN': {'clf__n_neighbors': [3,5,7], 'clf__weights': ['uniform', 'distance']},
    'SVM_linear': {'clf__C': [0.01, 0.03, 0.1, 0.3, 1, 3, 10]},
    'SVM_rbf': {'clf__C': [0.01, 0.03, 0.1, 0.3, 1, 3, 10], 'clf__gamma': ['scale','auto',0.1,0.01]}
}

#Generamos una nueva tabla con los hiperparamtros seleccionados anteriormente
results_grid = []

for name, pipeline in models.items():
    
    if name in ['SVM_linear', 'SVM_rbf']:
        search = RandomizedSearchCV(
            estimator=pipeline,
            param_distributions=param_grids[name],
            n_iter=20,
            scoring=scorer,
            cv=cv,
            n_jobs=-1,
            random_state=42,
            return_train_score=True
        )
    else:
        search = GridSearchCV(
            estimator=pipeline,
            param_grid=param_grids[name],
            scoring=scorer,
            cv=cv,
            n_jobs=-1,
            return_train_score=True
        )

    try:
        start = time.time()
        search.fit(X_train, y_train)
        elapsed = time.time() - start

        best_model = search.best_estimator_
        best_cv_score = search.best_score_
        y_pred_test = best_model.predict(X_test)
        test_f1 = f1_score(y_test, y_pred_test)
        best_index = search.best_index_
        mean_train_score = search.cv_results_.get('mean_train_score', [None])[best_index]

        results_grid.append({
            'Model': name,
            'Best Params': search.best_params_,
            'Best CV F1 (train, val)': round(best_cv_score, 4),
            'Mean Train F1 (CV)': (round(mean_train_score,4) if mean_train_score is not None else None),
            'Test F1': round(test_f1,4),
            'Search Time (s)': round(elapsed,2),
            'CV Std': round(search.cv_results_.get('std_test_score',[0])[best_index],4)
        })
    except Exception as e:
        print(f" No s'ha pogut entrenar {name}: {e}")
        results_grid.append({
            'Model': name,
            'Best Params': None,
            'Best CV F1 (train, val)': None,
            'Mean Train F1 (CV)': None,
            'Test F1': None,
            'Search Time (s)': None,
            'CV Std': None
        })

results_grid_df = pd.DataFrame(results_grid)

#Mostramos la tabla
results_grid_df['Score_final'] = results_grid_df['Best CV F1 (train, val)'] - results_grid_df['CV Std']
results_grid_df = results_grid_df.sort_values(['Score_final', 'Test F1'], ascending=False).reset_index(drop=True)
display(results_grid_df)
#Seleccionamos cual ha sido el mejor modelo en función de los resultados mostrados en la tabla previa
best_model_name = results_grid_df.iloc[0]['Model']
print(f"\n El mejor modelo ha sido: {best_model_name}")

# Entrenamos el modelo que nos ha salido mejor
best_pipeline = models[best_model_name]
best_pipeline.set_params(**results_grid_df.iloc[0]['Best Params'])
best_pipeline.fit(X_train, y_train)

# Predicciones sobre test
y_pred = best_pipeline.predict(X_test)
y_pred_proba = best_pipeline.predict_proba(X_test)[:,1] if hasattr(best_pipeline.named_steps['clf'], "predict_proba") else None

print(f"Best CV F1: {results_grid_df.iloc[0]['Best CV F1 (train, val)']}")
print(f"Test F1: {f1_score(y_test, y_pred):.4f}")
```

```python
"""
Comparativa de busquedas de hiperparámetros con el mismo tiempo
- Requiere sklearn >= 0.24 per HalvingRandomSearchCV
"""

import time
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV, GridSearchCV
from sklearn.experimental import enable_halving_search_cv 
from sklearn.model_selection import HalvingRandomSearchCV
from sklearn.metrics import f1_score, make_scorer

# Models + param grids (en función de lo de antes)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

# Definición Logística
model_name = 'LogisticRegression'
pipeline = Pipeline([('pre', preprocessor), ('clf', LogisticRegression(max_iter=2000, random_state=42))])
param_grid = {
    'clf__C': [0.01, 0.1, 1, 10, 100],
    'clf__solver': ['liblinear', 'lbfgs']
}

# Parámetros comunes
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scorer = make_scorer(f1_score)
time_budget_seconds = 60 * 30  

# Estiamos tiempo emdiano en función del candidato
t0 = time.time()
# Hacemos un cross_val_score amb pipeline 
from sklearn.model_selection import cross_val_score
_ = cross_val_score(pipeline.set_params(**{'clf__C':1, 'clf__solver':'lbfgs'}), X_train, y_train, cv=cv, scoring=scorer, n_jobs=-1)
elapsed_single = time.time() - t0
print(f"Tiempo estimado para una evaluación CV (s): {elapsed_single:.3f}")

# Estimación de cantidad de configuraciones que caben al presupuesto
estimated_configs = max(1, int(time_budget_seconds // elapsed_single))
print(f"Con nuestro presupuesto: {time_budget_seconds}s y tiempo de configuración {elapsed_single:.2f}s, estimamos ~{estimated_configs} configuraciones.")

# 2) Configuramos RandomizedSearchCV con n iteraciones basado en la estimación
n_iter = max(1, int(estimated_configs * 0.8))  # dejams 20% por si hubiera una sobrecarga
print(f"Usamos n_iter={n_iter} para RandomizedSearch.")

rand_search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_grid,
    n_iter=n_iter,
    scoring=scorer,
    cv=cv,
    n_jobs=-1,
    random_state=42,
    return_train_score=True
)

# 3) Configuramos HalvingRandomSearchCV con la misma estimación
halving = HalvingRandomSearchCV(
    estimator=pipeline,
    param_distributions=param_grid,
    n_candidates='exhaust',   # prueba tantas como permita el espacio
    factor=3,
    scoring=scorer,
    cv=cv,
    n_jobs=-1,
    random_state=42,
    return_train_score=True
)

# 4) (Opcional) Si el nombre total de combinaciones del grid es pequeño y cabe en el presupuesto, usamos GridSearch
from itertools import product
total_grid_combinations = np.prod([len(v) for v in param_grid.values()])
print(f"Total combinaciones Grid: {total_grid_combinations}")
# si total_grid_combinations < estimated_configs*1.2: cogemos grid
try_grid = total_grid_combinations <= estimated_configs * 1.1

# 5) Función para ejecutar con tiempo límite: lo hacemos con pruebas sequenciales y lo paramos si nos pasamos del presupuesto.
def run_with_budget(searcher, X_train, y_train, time_budget):
    start = time.time()
    searcher_start = time.time()
    searcher.fit(X_train, y_train)
    elapsed = time.time() - start
    return searcher, elapsed

results = []
#RandomizedSearch
t_start = time.time()
rand_search, rand_elapsed = run_with_budget(rand_search, X_train, y_train, time_budget_seconds)
print(f"RandomizedSearch ha acabado en {rand_elapsed:.1f}s. Best CV: {rand_search.best_score_:.4f}")
y_pred = rand_search.best_estimator_.predict(X_test)
test_f1 = f1_score(y_test, y_pred)
results.append(('RandomizedSearch', rand_search.best_params_, rand_search.best_score_, test_f1, rand_elapsed))

#HalvingRandomSearch
t0 = time.time()
halving, halving_elapsed = run_with_budget(halving, X_train, y_train, time_budget_seconds)
y_pred = halving.best_estimator_.predict(X_test)
test_f1 = f1_score(y_test, y_pred)
results.append(('HalvingRandomSearch', halving.best_params_, halving.best_score_, test_f1, halving_elapsed))
print(f"Halving ha acabado en {halving_elapsed:.1f}s. Best CV: {halving.best_score_:.4f}")

# (Opcional) GridSearch si cabe
if try_grid:
    grid = GridSearchCV(estimator=pipeline, param_grid=param_grid, scoring=scorer, cv=cv, n_jobs=-1, return_train_score=True)
    grid, grid_elapsed = run_with_budget(grid, X_train, y_train, time_budget_seconds)
    y_pred = grid.best_estimator_.predict(X_test)
    test_f1 = f1_score(y_test, y_pred)
    results.append(('GridSearch', grid.best_params_, grid.best_score_, test_f1, grid_elapsed))
    print(f"GridSearch ha acabado {grid_elapsed:.1f}s. Best_cv {grid.best_score_:.4f}")
else:
    print("GridSearch descartado porque el espacio de parámetros es demasiado grande para nuestro presupuesto.")

# Resumen
res_df = pd.DataFrame(results, columns=['Method', 'Best Params', 'Best CV F1', 'Test F1', 'Elapsed (s)'])
print("\nComparativa final:")
display(res_df)
res_df.to_csv('search_time_comparison_results.csv', index=False)
```

```python
# ============================================================
# 🔹 Correlación completa (numéricas + categóricas) con Survived
# ============================================================

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder

# 1️⃣ Cargar dataset
train = pd.read_csv("titanic/train.csv")

# 2️⃣ Crear variables nuevas (según vuestro código)
train['FamilySize'] = train['SibSp'] + train['Parch'] + 1
train['IsAlone'] = (train['FamilySize'] == 1).astype(int)

# Extraer el título del nombre
train['Title'] = train['Name'].str.extract(' ([A-Za-z]+)\.', expand=False)
train['Title'] = train['Title'].replace(['Lady', 'Countess','Capt', 'Col', 'Don', 'Dr', 
                                         'Major', 'Rev', 'Sir', 'Jonkheer', 'Dona'], 'Rare')
train['Title'] = train['Title'].replace({'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs'})

# 3️⃣ Rellenar valores nulos (como hacéis vosotros)
train['Age'] = train['Age'].fillna(train['Age'].median())
train['Embarked'] = train['Embarked'].fillna(train['Embarked'].mode()[0])

# 4️⃣ Seleccionar columnas relevantes
cols = ['Survived', 'Pclass', 'Sex', 'Age', 'Fare', 'Embarked', 'Title', 'FamilySize', 'IsAlone']
df = train[cols].copy()

# 5️⃣ One-Hot Encoding para variables categóricas
cat_cols = ['Sex', 'Embarked', 'Title']
encoder = OneHotEncoder(drop=None, sparse_output=False)
encoded = pd.DataFrame(encoder.fit_transform(df[cat_cols]), 
                       columns=encoder.get_feature_names_out(cat_cols),
                       index=df.index)

df_encoded = pd.concat([df.drop(columns=cat_cols), encoded], axis=1)

# 6️⃣ Calcular correlación con Survived
corr = df_encoded.corr(numeric_only=True)['Survived'].sort_values(ascending=False)

# 7️⃣ Mostrar las correlaciones visualmente
plt.figure(figsize=(10,6))
sns.barplot(x=corr.index, y=corr.values, palette="coolwarm")
plt.title("Correlación de todas las variables con la supervivencia", fontsize=14)
plt.ylabel("Coeficiente de correlación", fontsize=12)
plt.xticks(rotation=90)
plt.axhline(0, color="black", linewidth=1)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()

# 8️⃣ Mostrar las 10 más correlacionadas
print("🔝 Variables más correlacionadas con Survived:")
print(corr.head(10))
```
