```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
# Extraer título de 'Name' usando raw string para la expresión regular
train['Title'] = train['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)
test['Title'] = test['Name'].str.extract(r' ([A-Za-z]+)\.', expand=False)

# Tamaño de la familia
train['FamilySize'] = train['SibSp'] + train['Parch'] + 1
test['FamilySize'] = test['SibSp'] + test['Parch'] + 1

# Variable binaria: viaja solo o no
train['IsAlone'] = (train['FamilySize'] == 1).astype(int)
test['IsAlone'] = (test['FamilySize'] == 1).astype(int)

# =========================
# 4️⃣ Selección de columnas
# =========================
numeric_features = ['Age', 'Fare', 'FamilySize']
categorical_features = ['Pclass', 'Sex', 'Embarked', 'Title', 'IsAlone']

# =========================
# 5️⃣ Preprocesado con pipeline
# =========================
# Para variables numéricas: rellenar NaN con mediana y normalizar
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())  # normalización: útil para modelos como SVM o KNN
])

# Para variables categóricas: rellenar NaN y One-Hot Encoding
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

# Combinar transformadores
preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_transformer, numeric_features),
    ('cat', categorical_transformer, categorical_features)
])

# =========================
# 6️⃣ Aplicar preprocesado
# =========================
X_train = preprocessor.fit_transform(train)
y_train = train['Survived']

X_test = preprocessor.transform(test)

print("\nPreprocesado completado. Datos listos para modelos.")
print("Dimensiones X_train:", X_train.shape)
print(X_train[:5])  # muestra las primeras 5 filas
```

```python
# ======================================
# 3️⃣ Mètric Selection — Regressió Logística
# ======================================

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    average_precision_score,
    precision_recall_curve,
    roc_curve,
    roc_auc_score,
    classification_report
)
import matplotlib.pyplot as plt

# --------------------------------------
# Entrenament del model
# --------------------------------------
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)

# Prediccions
y_pred = log_reg.predict(X_test)
y_pred_proba = log_reg.predict_proba(X_test)[:, 1]

# --------------------------------------
# 1️⃣ Càlcul de mètriques principals
# --------------------------------------
acc = accuracy_score(y_train, log_reg.predict(X_train))
f1 = f1_score(y_train, log_reg.predict(X_train))
avg_prec = average_precision_score(y_train, log_reg.predict_proba(X_train)[:, 1])

print("Accuracy:", round(acc, 3))
print("F1 Score:", round(f1, 3))
print("Average Precision:", round(avg_prec, 3))

# --------------------------------------
# 2️⃣ Corbes de rendiment
# --------------------------------------

def plot_precision_recall_curve(y_true, y_scores):
    precision, recall, _ = precision_recall_curve(y_true, y_scores)
    plt.figure(figsize=(6,4))
    plt.plot(recall, precision, lw=2)
    plt.title("Precision-Recall Curve")
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.grid(True)
    plt.show()

def plot_roc_curve(y_true, y_scores):
    fpr, tpr, _ = roc_curve(y_true, y_scores)
    auc = roc_auc_score(y_true, y_scores)
    plt.figure(figsize=(6,4))
    plt.plot(fpr, tpr, lw=2, label=f"AUC = {auc:.3f}")
    plt.plot([0,1],[0,1],'k--')
    plt.title("ROC Curve")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.legend()
    plt.grid(True)
    plt.show()

# Mostrem les corbes
plot_precision_recall_curve(y_train, log_reg.predict_proba(X_train)[:, 1])
plot_roc_curve(y_train, log_reg.predict_proba(X_train)[:, 1])

# --------------------------------------
# 3️⃣ Classification report
# --------------------------------------
print("\nCLASSIFICATION REPORT:")
print(classification_report(y_train, log_reg.predict(X_train)))
```

```python
# ============================================================
# MODEL COMPARISON + GRID SEARCH (amb imputació de NaN)
# ============================================================

import time
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.metrics import f1_score, make_scorer
from sklearn.impute import SimpleImputer
from sklearn.datasets import fetch_openml

# ------------------------------------------------------------
# 0️⃣ Comprovació i preparació de les dades
# ------------------------------------------------------------
try:
    X
    y
except NameError:
    print("⚙️ Carregant dataset Titanic des d'OpenML...")
    data = fetch_openml("titanic", version=1, as_frame=True)

    # Dades i objectiu
    X = data.data.copy()
    y = data.target.copy()

    # Correcció del vector objectiu (important)
    if y.dtype == object or y.dtype.name == 'category':
        y = y.map({'yes': 1, 'no': 0, '1': 1, '0': 0}).astype(float)

    y = pd.to_numeric(y, errors='coerce')
    valid_idx = y.dropna().index
    X = X.loc[valid_idx].reset_index(drop=True)
    y = y.loc[valid_idx].astype(int).reset_index(drop=True)

# Comprovació de classes
print("\n🔍 Distribució de classes a y:")
print(y.value_counts(normalize=True))

# ------------------------------------------------------------
# 1️⃣ Separació train/test i detecció de tipus de columnes
# ------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

cat_cols = X_train.select_dtypes(include=['object', 'category']).columns
num_cols = X_train.select_dtypes(exclude=['object', 'category']).columns

print(f"\nColumnes numèriques: {list(num_cols)}")
print(f"Columnes categòriques: {list(cat_cols)}")

# ------------------------------------------------------------
# 2️⃣ Preprocessador complet (amb imputació)
# ------------------------------------------------------------
numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),   # omple NaN amb mediana
    ('scaler', StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),  # omple NaN amb valor més freqüent
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer([
    ('num', numeric_transformer, num_cols),
    ('cat', categorical_transformer, cat_cols)
])

# ------------------------------------------------------------
# 3️⃣ Definició de models
# ------------------------------------------------------------
models = {
    'LogisticRegression': Pipeline([
        ('pre', preprocessor),
        ('clf', LogisticRegression(max_iter=2000, random_state=42))
    ]),
    'KNN': Pipeline([
        ('pre', preprocessor),
        ('clf', KNeighborsClassifier())
    ]),
    'SVM_linear': Pipeline([
        ('pre', preprocessor),
        ('clf', SVC(kernel='linear', probability=True, random_state=42))
    ]),
    'SVM_rbf': Pipeline([
        ('pre', preprocessor),
        ('clf', SVC(kernel='rbf', probability=True, random_state=42))
    ])
}

# ------------------------------------------------------------
# 4️⃣ Validació creuada i mètrica
# ------------------------------------------------------------
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scorer = make_scorer(f1_score)

# ------------------------------------------------------------
# 5️⃣ Comparació de models amb hiperparàmetres per defecte
# ------------------------------------------------------------
results_default = []

for name, pipeline in models.items():
    print(f"\n🔹 Avaluant model: {name}")
    try:
        start = time.time()
        scores = cross_val_score(pipeline, X_train, y_train, cv=cv, scoring=scorer, n_jobs=-1, error_score='raise')
        elapsed = time.time() - start
        results_default.append({
            'Model': name,
            'Mean CV F1 (train)': np.mean(scores),
            'Std CV F1': np.std(scores),
            'Train Time (s) (CV)': round(elapsed, 2)
        })
    except Exception as e:
        print(f"⚠️ Error en model {name}: {e}")
        results_default.append({
            'Model': name,
            'Mean CV F1 (train)': None,
            'Std CV F1': None,
            'Train Time (s) (CV)': None
        })

results_default_df = pd.DataFrame(results_default).sort_values('Mean CV F1 (train)', ascending=False).reset_index(drop=True)
print("\n📊 Resultats inicials (hiperparàmetres per defecte):")
display(results_default_df)

# ------------------------------------------------------------
# 6️⃣ Quadrícules d'hiperparàmetres per Grid Search
# ------------------------------------------------------------
param_grids = {
    'LogisticRegression': {
        'clf__C': [0.01, 0.1, 1, 10],
        'clf__solver': ['liblinear', 'lbfgs']
    },
    'KNN': {
        'clf__n_neighbors': [3, 5, 7, 9, 11],
        'clf__weights': ['uniform', 'distance']
    },
    'SVM_linear': {
        'clf__C': [0.01, 0.1, 1, 10]
    },
    'SVM_rbf': {
        'clf__C': [0.01, 0.1, 1, 10],
        'clf__gamma': ['scale', 'auto', 0.1, 0.01]
    }
}

# ------------------------------------------------------------
# 7️⃣ Grid Search per a cada model
# ------------------------------------------------------------
results_grid = []

for name, pipeline in models.items():
    print(f"\n🔍 Cercant hiperparàmetres per: {name}")
    grid = GridSearchCV(
        estimator=pipeline,
        param_grid=param_grids[name],
        scoring=scorer,
        cv=cv,
        n_jobs=-1,
        verbose=0,
        return_train_score=True,
        error_score='raise'
    )

    try:
        start = time.time()
        grid.fit(X_train, y_train)
        elapsed = time.time() - start

        best_model = grid.best_estimator_
        best_cv_score = grid.best_score_

        # Predicció al test
        y_pred_test = best_model.predict(X_test)
        test_f1 = f1_score(y_test, y_pred_test)

        best_index = grid.best_index_
        mean_train_score = grid.cv_results_.get('mean_train_score', [None])[best_index]

        results_grid.append({
            'Model': name,
            'Best Params': grid.best_params_,
            'Best CV F1 (train, val)': round(best_cv_score, 4),
            'Mean Train F1 (CV)': (round(mean_train_score, 4) if mean_train_score is not None else None),
            'Test F1': round(test_f1, 4),
            'Grid Search Time (s)': round(elapsed, 2)
        })
    except Exception as e:
        print(f"⚠️ Error en Grid Search per {name}: {e}")
        results_grid.append({
            'Model': name,
            'Best Params': None,
            'Best CV F1 (train, val)': None,
            'Mean Train F1 (CV)': None,
            'Test F1': None,
            'Grid Search Time (s)': None
        })

results_grid_df = pd.DataFrame(results_grid).sort_values('Best CV F1 (train, val)', ascending=False).reset_index(drop=True)
print("\n📈 Resultats de Grid Search:")
display(results_grid_df)

# ------------------------------------------------------------
# 8️⃣ Taula resum final
# ------------------------------------------------------------
def get_default_score(name):
    row = results_default_df[results_default_df['Model'] == name]
    if not row.empty:
        return float(row['Mean CV F1 (train)'].values[0])
    return None

summary = results_grid_df.copy()
summary['Default CV F1 (train)'] = summary['Model'].apply(get_default_score)
summary = summary[['Model', 'Default CV F1 (train)', 'Best Params',
                   'Best CV F1 (train, val)', 'Mean Train F1 (CV)',
                   'Test F1', 'Grid Search Time (s)']]

print("\n📋 Taula resum final:")
display(summary)

summary.to_csv("model_selection_summary.csv", index=False)
print("\n💾 Resultats desats a model_selection_summary.csv")
```

```python
"""
Comparativa de cerques d'hiperparàmetres amb el mateix temps (estimació)
- Requereix sklearn >= 0.24 per HalvingRandomSearchCV
- Assumim que X_train, X_test, y_train, y_test, preprocessor estan definits (igual que abans)
"""

import time
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV, GridSearchCV
from sklearn.experimental import enable_halving_search_cv  # noqa
from sklearn.model_selection import HalvingRandomSearchCV
from sklearn.metrics import f1_score, make_scorer

# Models + param grids (pots modificar segons el que vas fer abans)
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

# Definició (exemple per 1 model: LogisticRegression). Pots repetir per altres models.
model_name = 'LogisticRegression'
pipeline = Pipeline([('pre', preprocessor), ('clf', LogisticRegression(max_iter=2000, random_state=42))])
param_grid = {
    'clf__C': [0.01, 0.1, 1, 10, 100],
    'clf__solver': ['liblinear', 'lbfgs']
}

# Paràmetres comuns
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scorer = make_scorer(f1_score)
time_budget_seconds = 60 * 30  # p.ex. 0.5h = 1800s; canvia a 3600 per 1h

# 1) Estimar temps per candidat (mitjana)
# Fem una prova amb una única configuració (potser la per defecte) per mesurar temps d'una avaluació
print("Estimant temps per candidat (1 fit+CV)...")
t0 = time.time()
# fer un cross_val_score amb pipeline una vegada
from sklearn.model_selection import cross_val_score
_ = cross_val_score(pipeline.set_params(**{'clf__C':1, 'clf__solver':'lbfgs'}), X_train, y_train, cv=cv, scoring=scorer, n_jobs=-1)
elapsed_single = time.time() - t0
print(f"Temps estimat per a una avaluació CV (s): {elapsed_single:.3f}")

# Estimació de quantitat de configuracions que caben al pressupost
# Tenir en compte que GridSearch avaluarà cada combinació; RandomizedSearch avaluarà n_iter; Halving pot fer menys avaluacions però amb iteracions internes.
estimated_configs = max(1, int(time_budget_seconds // elapsed_single))
print(f"Amb pressupost {time_budget_seconds}s i temps per configuració {elapsed_single:.2f}s, estimem ~{estimated_configs} configuracions.")

# 2) Configurar RandomizedSearchCV amb n_iter basat en l'estimació (deixar marge)
n_iter = max(1, int(estimated_configs * 0.8))  # deixar 20% marge per sobrecàrrega
print(f"Usarem n_iter={n_iter} per RandomizedSearch.")

rand_search = RandomizedSearchCV(
    estimator=pipeline,
    param_distributions=param_grid,
    n_iter=n_iter,
    scoring=scorer,
    cv=cv,
    n_jobs=-1,
    random_state=42,
    return_train_score=True
)

# 3) Configurar HalvingRandomSearchCV amb la mateixa estimació (els paràmetres poden ajustar-se)
halving = HalvingRandomSearchCV(
    estimator=pipeline,
    param_distributions=param_grid,
    n_candidates='exhaust',   # prova tantes com permeti l'espai (si capitals) o 'all' depenent versió
    factor=3,
    scoring=scorer,
    cv=cv,
    n_jobs=-1,
    random_state=42,
    return_train_score=True
)

# 4) (Opcional) si el nombre total de combinacions del grid és petit i càpiga al pressupost, fem GridSearch
from itertools import product
total_grid_combinations = np.prod([len(v) for v in param_grid.values()])
print(f"Total combinacions Grid: {total_grid_combinations}")
# si total_grid_combinations < estimated_configs*1.2: llavors podem provar Grid
try_grid = total_grid_combinations <= estimated_configs * 1.1

# 5) Funció per executar amb temps límit: fem proves seqüencials i aturem si excedim budget (seguretat)
def run_with_budget(searcher, X_train, y_train, time_budget):
    start = time.time()
    searcher_start = time.time()
    searcher.fit(X_train, y_train)
    elapsed = time.time() - start
    return searcher, elapsed

results = []

# Run RandomizedSearch
print("Executant RandomizedSearchCV ...")
t_start = time.time()
rand_search, rand_elapsed = run_with_budget(rand_search, X_train, y_train, time_budget_seconds)
print(f"RandomizedSearch acabat en {rand_elapsed:.1f}s. Best CV: {rand_search.best_score_:.4f}")
y_pred = rand_search.best_estimator_.predict(X_test)
test_f1 = f1_score(y_test, y_pred)
results.append(('RandomizedSearch', rand_search.best_params_, rand_search.best_score_, test_f1, rand_elapsed))

# Run HalvingRandomSearchCV
print("Executant HalvingRandomSearchCV ...")
t0 = time.time()
halving, halving_elapsed = run_with_budget(halving, X_train, y_train, time_budget_seconds)
y_pred = halving.best_estimator_.predict(X_test)
test_f1 = f1_score(y_test, y_pred)
results.append(('HalvingRandomSearch', halving.best_params_, halving.best_score_, test_f1, halving_elapsed))
print(f"Halving acabat: best_cv {halving.best_score_:.4f} en {halving_elapsed:.1f}s")

# (Opcional) GridSearch si càpiga
if try_grid:
    print("Executant GridSearchCV (sota pressupost estimat)...")
    grid = GridSearchCV(estimator=pipeline, param_grid=param_grid, scoring=scorer, cv=cv, n_jobs=-1, return_train_score=True)
    grid, grid_elapsed = run_with_budget(grid, X_train, y_train, time_budget_seconds)
    y_pred = grid.best_estimator_.predict(X_test)
    test_f1 = f1_score(y_test, y_pred)
    results.append(('GridSearch', grid.best_params_, grid.best_score_, test_f1, grid_elapsed))
    print(f"GridSearch acabat: best_cv {grid.best_score_:.4f} en {grid_elapsed:.1f}s")
else:
    print("GridSearch descartat perquè l'espai de paràmetres és massa gran per al pressupost estimat.")

# Mostrar resum
res_df = pd.DataFrame(results, columns=['Method', 'Best Params', 'Best CV F1', 'Test F1', 'Elapsed (s)'])
print("\nComparativa final (mateix pressupost aproximat):")
display(res_df)
res_df.to_csv('search_time_comparison_results.csv', index=False)
print("Resultats desats a search_time_comparison_results.csv")
```
